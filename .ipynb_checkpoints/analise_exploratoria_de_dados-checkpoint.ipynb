# Análise Exploratória de Dados

A Análise Exploratória de Dados (EDA) é o processo inicial de investigação em que resumimos as principais características de um conjunto de dados, geralmente usando métodos de visualização de dados. O objetivo é entender a estrutura, identificar padrões, detectar anomalias e testar hipóteses. A EDA é um passo crucial que guia a modelagem, ajudando a tomar decisões informadas sobre como pré-processar os dados e quais algoritmos usar.

## Funções

A definição de funções neste notebook tem como objetivo principal tornar o código mais modular, organizado e reutilizável. Em vez de repetir blocos de código, criamos funções que executam tarefas específicas, como a criação de visualizações padronizadas. Isso melhora a clareza do código e facilita a manutenção, permitindo que alterações sejam feitas em um único local e aplicadas em todo o projeto.

````python
def analyze_categorical_variable(df, column_name):
    """
    Realiza uma análise completa de uma variável categórica, exibindo
    tabelas de métricas e contagens, e um gráfico de distribuição.

    Args:
        df (pd.DataFrame): O DataFrame de dados.
        column_name (str): O nome da coluna a ser analisada.
    """
    if column_name not in df.columns:
        print(f"Erro: A coluna '{column_name}' não existe no DataFrame.")
        return

    print(f"Análise da Variável '{column_name}'")
    data_series = df[column_name]
    total_rows = len(data_series)
    
    # 1. Calcula e exibe as métricas de resumo
    distinct_count = data_series.nunique(dropna=False)
    missing_count = data_series.isnull().sum()
    missing_pct = (missing_count / total_rows) * 100
    
    zeros_count = 0
    negative_count = 0
    mean_val, min_val, max_val = "N/A", "N/A", "N/A"
    
    if pd.api.types.is_numeric_dtype(data_series):
        zeros_count = (data_series == 0).sum()
        negative_count = (data_series < 0).sum()
        mean_val = data_series.mean()
        min_val = data_series.min()
        max_val = data_series.max()

    zeros_pct = (zeros_count / total_rows) * 100
    negative_pct = (negative_count / total_rows) * 100

    summary_data = {
        'Métrica': ['Distinct', 'Distinct (%)', 'Missing', 'Missing (%)', 
                    'Zeros', 'Zeros (%)', 'Negative', 'Negative (%)',
                    'Mean', 'Minimum', 'Maximum'],
        'Valor': [distinct_count, f"{distinct_count / total_rows * 100:.2f}%", 
                  missing_count, f"{missing_pct:.2f}%", 
                  zeros_count, f"{zeros_pct:.2f}%", 
                  negative_count, f"{negative_pct:.2f}%",
                  f"{mean_val:.2f}" if isinstance(mean_val, float) else mean_val, 
                  min_val, max_val]
    }
    summary_df = pd.DataFrame(summary_data)
    
    print("\n--- Resumo das Métricas ---")
    display(summary_df)

    # 2. Calcula e exibe a tabela de contagem por categoria
    count_df = data_series.value_counts(dropna=False).to_frame('Contagem')
    count_df['Proporção'] = count_df['Contagem'] / total_rows
    
    print("\n--- Contagem e Proporção por Categoria ---")
    display(count_df)

    print("\n--- Gráfico de Distribuição ---")
    # 3. Gera o gráfico de distribuição
    plt.figure(figsize=(10, 6))
    ax = sns.countplot(
        x=column_name,
        data=df,
        order=data_series.value_counts().index,
        palette='viridis',
        hue=column_name,  
        legend=False   
    )
    plt.title(f'Distribuição de {column_name}', fontsize=16)
    plt.xlabel('Categoria', fontsize=12)
    plt.ylabel('Contagem', fontsize=12)
    plt.xticks(rotation=45, ha='right')

    # Adiciona os valores em cada barra
    for p in ax.patches:
        height = p.get_height()
        ax.annotate(f'{int(height)}',
                    (p.get_x() + p.get_width() / 2., height),
                    ha='center', va='bottom',
                    fontsize=10,
                    xytext=(0, 5),
                    textcoords='offset points')
                    
    plt.show()

# Exemplo de uso
# Suponha que 'df' é o seu DataFrame e 'cbc_sexo' é a variável que você quer analisar.
# Basta chamar a função com o nome da coluna:
# analyze_categorical_variable(df, 'cbc_sexo')
````

## Variáveis Categóricas/Ordinais/Numéricas


Para realizar uma análise de dados eficaz, é fundamental entender o tipo de cada variável no nosso conjunto. As variáveis categóricas representam grupos ou categorias sem uma ordem intrínseca (por exemplo, ```cbc_sexo```). As variáveis ordinais também são categorias, mas possuem uma ordem lógica (por exemplo, ```ord_nivel_ensino```). Já as variáveis numéricas representam quantidades (como ```num_idade_morador``` ou ```num_altura```). A função ```analyze_categorical_variable``` é uma ferramenta poderosa para a análise desses tipos de variáveis. Ela gera automaticamente um resumo completo, incluindo contagens de valores e porcentagens, e visualizações em gráficos de barras. Isso nos permite avaliar a distribuição, a proporção de dados faltantes ou vazios, e a frequência de cada categoria, fornecendo uma base sólida para a limpeza e a preparação dos dados antes da modelagem.

````python
analyze_categorical_variable(df, 'var_alvo')
````

### Distribuição da Variável Alvo pela Idade

### Gráfico Radar da Variável Alvo pelas Variáveis relacionadas com Perfis de Dificuldade

````python
vars_dificuldade = [
    'ord_dif_comer', 'ord_dif_banho', 'ord_dif_banheiro', 'ord_dif_vestir',
    'ord_dif_andar_casa', 'ord_dif_cama', 'ord_dif_cadeira', 'ord_dif_compras',
    'ord_dif_financas', 'ord_dif_ir_medico', 'ord_dif_transporte'
]

mapeamento_dificuldade = {
    'Não informado': 0,
    'Não tem dificuldade': 1,
    'Tem pequena dificuldade': 2,
    'Tem grande dificuldade': 3,
    'Não consegue': 4
}

df_graph = df.copy()
````

````python
# 2. Aplique o mapeamento a cada coluna
for col in vars_dificuldade:
    df_graph[col] = df_graph[col].map(mapeamento_dificuldade).fillna(0)
````

````python
# Calcule a média de cada variável para cada grupo da variável alvo
# Substitua 'df' pelo seu DataFrame
df_radar = df_graph.groupby('var_alvo')[vars_dificuldade].mean().reset_index()

# Extraia os valores para os dois grupos (var_alvo = 0 e var_alvo = 1)
valores_alvo_0 = df_radar.loc[df_radar['var_alvo'] == 0, vars_dificuldade].values[0]
valores_alvo_1 = df_radar.loc[df_radar['var_alvo'] == 1, vars_dificuldade].values[0]

# O número de variáveis que serão os eixos do gráfico
num_vars = len(vars_dificuldade)

# Calcule os ângulos para cada eixo
angles = np.linspace(0, 2 * np.pi, num_vars, endpoint=False).tolist()

# Para fechar o círculo do gráfico, repita o primeiro valor no final
valores_alvo_0 = np.concatenate((valores_alvo_0, [valores_alvo_0[0]]))
valores_alvo_1 = np.concatenate((valores_alvo_1, [valores_alvo_1[0]]))
angles = np.concatenate((angles, [angles[0]]))
````

````python
# **3. Criação e Personalização do Gráfico de Radar**
fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))

ax.set_facecolor('whitesmoke')

# Plote os perfis dos dois grupos
ax.plot(angles, valores_alvo_0, linewidth=2, linestyle='solid', label='var_alvo = 0', color='skyblue')
ax.fill(angles, valores_alvo_0, color='skyblue', alpha=0.25)
ax.plot(angles, valores_alvo_1, linewidth=2, linestyle='solid', label='var_alvo = 1', color='salmon')
ax.fill(angles, valores_alvo_1, color='salmon', alpha=0.25)

# Adicione os rótulos das variáveis e os ticks numéricos do eixo Y
ax.set_xticks(angles[:-1])
ax.set_xticklabels(vars_dificuldade, fontsize=10)
ax.set_yticks(np.arange(5))
ax.set_yticklabels(['0', '1', '2', '3', '4'])
ax.set_ylabel('Nível de Dificuldade', fontsize=12, labelpad=25)

# Adicione a legenda para os perfis
ax.legend(loc='upper right', bbox_to_anchor=(0.1, 0.1), title='Variável Alvo')

# **--- Adiciona a legenda de mapeamento ---**
# Formata o texto da legenda
texto_legenda = 'Mapeamento:\n => ' + '\n => '.join([f'{v}: {k}' for k, v in mapeamento_dificuldade.items()])

# Adiciona um bloco de texto no gráfico
plt.figtext(0.992, 0.79, texto_legenda, fontsize=10, ha='left', va='top', bbox=dict(facecolor='white', alpha=0.5, edgecolor='gray'))

# Adicione título e legenda
ax.set_title('Perfil de Dificuldades x Variável Alvo', fontsize=16, y=1.1)
ax.legend(bbox_to_anchor=(1.35, 0.99))

# Exiba o gráfico
plt.show()
````

### Catplot do tipo caixa da Variável Alvo pelas Variáveis relacionadas com Frequência de Alimentação

````python
# As variáveis de frequência que você quer analisar
vars_frequencia = [
    'num_freq_verduras_semana', 'num_freq_carne_vermelha', 'num_freq_frango',
    'num_freq_peixe', 'num_freq_frutas',
    'num_freq_doces_semana', 'num_freq_refeicao_lanches'
]

# Certifique-se de que os valores nulos sejam tratados.
# Se `null` significa "Não aplicável", o catplot irá ignorá-los automaticamente.

# 1. Reestruture o DataFrame usando .melt()
# A variável 'var_alvo' será a sua variável de identificação (id_vars)
df_longo = df.melt(
    id_vars=['var_alvo'],
    value_vars=vars_frequencia,
    var_name='tipo_de_alimento',
    value_name='frequencia_consumo'
)
````

````python
# 2. Crie a plotagem usando sns.catplot
sns.set_style("whitegrid")
g = sns.catplot(
    data=df_longo,
    x='tipo_de_alimento',
    y='frequencia_consumo',
    hue='var_alvo',
    kind='box',
    height=6,
    aspect=2.5,
    palette='Set2'
)
````

### Catplot do tipo barra da Variável Alvo pelas Variáveis relacionadas com Frequência de Alimentação
